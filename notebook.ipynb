```python
import numpy as np
from collections import defaultdict
from typing import List
```

```python
strs=["act","pots","tops","cat","stop","hat"]
numystringarray = np.array(strs)
print (numystringarray)
```

```python
result_map = {}
for s in numystringarray:
    print(result_map)
    sorted_s = "".join(sorted(s))
    if sorted_s not in result_map:
        result_map[sorted_s] = []
    result_map[sorted_s].append(s)
```

```python
print(list(result_map.values()))
```
